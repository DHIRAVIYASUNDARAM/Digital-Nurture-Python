```python
class Employee:
    def __init__(self, name):
        self.name = name
        self.salary = 0

    def set_salary(self, salary):
        if salary > 0:
            self.salary = salary
        return self

    def apply_raise(self, percent):
        if percent > 0:
            self.salary += self.salary * (percent / 100)
        return self

    def display(self):
        print(f"{self.name}'s Salary: {self.salary:.2f}")
        return self

emp = Employee("John")
emp.set_salary(50000).apply_raise(10).display()
```

```python
import configparser

class Config:
    pass

class DatabaseConfig(Config):

    def load(self, filename):
        config = configparser.ConfigParser()

        files_read = config.read(filename)

        if not files_read:
            print("Error: File not found")
            return

        if "DATABASE" not in config:
            print("Error: DATABASE section not found")
            return

        required = ["host", "user", "password"]

        for key in required:
            if key not in config["DATABASE"]:
                print(f"Missing key: {key}")
                return

        print("Database Configuration Loaded Successfully")
        print("\nDatabase Settings:")
        print("Host     :", config["DATABASE"]["host"])
        print("User     :", config["DATABASE"]["user"])
        print("Password :", config["DATABASE"]["password"])

db = DatabaseConfig()
db.load("db.ini")
```

```python
def calculate(a, b, op):
    try:
        if op == "+":
            return a + b
        elif op == "-":
            return a - b
        elif op == "*":
            return a * b
        elif op == "/":
            return a / b
        else:
            return "Invalid Operator"

    except ZeroDivisionError:
        return "Cannot divide by zero"

try:
    a = float(input("Enter First Number: "))
    b = float(input("Enter Second Number: "))
    op = input("Enter Operator (+,-,*,/): ")

    print("Result:", calculate(a, b, op))

except ValueError:
    print("Invalid Number")
```

```python
class CartItem:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, item):
        self.items.append(item)

    def remove_item(self, name):
        self.items = [item for item in self.items if item.name != name]

    def total(self):
        return sum(item.price * item.quantity for item in self.items)

cart = ShoppingCart()

cart.add_item(CartItem("Laptop", 50000, 1))
cart.add_item(CartItem("Mouse", 500, 2))

subtotal = cart.total()
gst = subtotal * 0.18

print("Subtotal:", subtotal)
print("GST:", gst)
print("Final Total:", subtotal + gst)
```

```python
from tabulate import tabulate

class Converter:
    def c_to_f(self, c):
        return (c * 9/5) + 32

    def c_to_k(self, c):
        return c + 273.15

temp = float(input("Enter Celsius: "))

converter = Converter()

table = [
    ["Celsius", temp],
    ["Fahrenheit", round(converter.c_to_f(temp), 2)],
    ["Kelvin", round(converter.c_to_k(temp), 2)]
]

print(tabulate(table, headers=["Unit", "Value"]))
```

```python
import hashlib

class URLShortener:
    def __init__(self):
        self.url_map = {}

    def shorten_url(self, url):
        short_code = hashlib.md5(url.encode()).hexdigest()[:6]
        self.url_map[short_code] = url
        return short_code

    def get_original_url(self, short_code):
        return self.url_map.get(short_code, "URL not found")

shortener = URLShortener()

url = "https://www.google.com"
short_code = shortener.shorten_url(url)

print("Original URL:", url)
print("Short Code:", short_code)
print("Retrieved URL:", shortener.get_original_url(short_code))
```

```python
import json

students = {}

def add_grade(student, grade):
    if not (0 <= grade <= 100):
        print("Invalid Grade")
        return

    students.setdefault(student, []).append(grade)

def calculate_gpa(student):
    grades = students.get(student, [])
    if not grades:
        return 0

    return round(sum(grades) / len(grades), 2)

def save_data():
    with open("grades.json", "w") as file:
        json.dump(students, file)

def load_data():
    try:
        with open("grades.json", "r") as file:
            return json.load(file)
    except FileNotFoundError:
        return {}

add_grade("John", 85)
add_grade("John", 90)
add_grade("Alice", 78)

save_data()

print("John GPA:", calculate_gpa("John"))

loaded = load_data()
class_avg = sum(
    sum(grades) / len(grades)
    for grades in loaded.values()
) / len(loaded)

print("Class Average:", round(class_avg, 2))
```

```python
from datetime import datetime

class Task:
    def __init__(self, name, due_date, priority):
        self.name = name
        self.due_date = datetime.strptime(due_date, "%Y-%m-%d")
        self.priority = priority

    def __str__(self):
        return f"{self.name} | Due: {self.due_date.date()} | Priority: {self.priority}"

tasks = [
    Task("Assignment", "2026-06-10", 1),
    Task("Project", "2026-06-05", 2),
    Task("Exam Prep", "2026-06-08", 1)
]

tasks.sort(key=lambda task: task.due_date)

print("Task Schedule:")
for task in tasks:
    print(task)

today = datetime.now()

print("\nOverdue Tasks:")
for task in tasks:
    if task.due_date < today:
        print(task)
```

```python
class Product:
    def __init__(self, name, stock):
        self.name = name
        self.stock = stock

class Perishable(Product):
    pass

class Electronics(Product):
    pass

inventory = {
    "Milk": Perishable("Milk", 5),
    "Laptop": Electronics("Laptop", 25),
    "Bread": Perishable("Bread", 3),
    "Phone": Electronics("Phone", 15)
}

low_stock = set()

for product in inventory.values():
    if product.stock < 10:
        low_stock.add(product.name)

print("Inventory Summary:")
for name, product in inventory.items():
    print(f"{name}: {product.stock} units")

print("\nLow Stock Alerts:")
for item in low_stock:
    print(item)
```

```python
import matplotlib.pyplot as plt

class Category:
    def __init__(self, name, limit):
        self.name = name
        self.limit = limit
        self.spent = 0

    def add_expense(self, amount):
        self.spent += amount

    def check_budget(self):
        if self.spent > self.limit:
            print(f"Alert: {self.name} exceeded budget!")

categories = {
    "Food": Category("Food", 5000),
    "Travel": Category("Travel", 3000),
    "Entertainment": Category("Entertainment", 2000)
}

while True:
    category = input("Enter Category (or 'done'): ")

    if category.lower() == "done":
        break

    if category in categories:
        amount = float(input("Enter Expense Amount: "))
        categories[category].add_expense(amount)
        categories[category].check_budget()

labels = []
values = []

for category in categories.values():
    labels.append(category.name)
    values.append(category.spent)

plt.pie(values, labels=labels, autopct="%1.1f%%")
plt.title("Monthly Budget Usage")
plt.show()
```
